8. **Create Financial Metrics**

```python
Sales = Sales.merge(Products[["Product_ID", "Unit_Cost"]], on = "Product_ID", how = "left")
```

```python
# Keep only Delivered orders for financial calculations
delivered = Sales["Delivery_Status"].astype(str).str.strip().str.lower() == "delivered"
```

```python
# Gross Revenue
Sales["Gross_Revenue"] = np.where(
    delivered,
    Sales["Quantity"] * Sales["Unit_Price"],
    0
)
```

```python
# Discount Amount
Sales["Discount_Amount"] = np.where(
    delivered,
    round(Sales["Gross_Revenue"] * Sales["Discount_Pct"] / 100, 2),
    0
)
```

```python
# Net Revenue
Sales["Net_Revenue"] = np.where(
    delivered,
    Sales["Gross_Revenue"] - Sales["Discount_Amount"],
    0
)
```

```python
# Total Cost
Sales["Total_Cost"] = np.where(
    delivered,
    Sales["Quantity"] * Sales["Unit_Cost"],
    0
)
```

```python
# Gross Profit
Sales["Gross_Profit"] = np.where(
    delivered,
    Sales["Net_Revenue"] - Sales["Total_Cost"],
    0
)
```

```python
# Profit Margin
Sales["Profit_Margin"] = np.where(
    Sales["Net_Revenue"] != 0,
    round((Sales["Gross_Profit"] / Sales["Net_Revenue"]) * 100, 2),
    0
)
```

**(xix).Budget vs Actual_Revenue**

```python
# Change into same datatype
Sales['Order_Date'] = pd.to_datetime(Sales['Order_Date'])
Budget['Date'] = pd.to_datetime(Budget['Date'])
```

```python
Budget['Month'] = Budget['Date'].dt.strftime('%b')
```

```python
actual_revenue = (
    Sales.groupby(['Month', 'Product_ID'])['Net_Revenue']
    .sum().reset_index(name='Actual_Revenue'))
```

```python
actual_revenue["Actual_Revenue"] = actual_revenue["Actual_Revenue"].astype(int)
```

```python
monthly_budget = (
    Budget.groupby(['Month', 'Product_ID'])['Budget_Revenue']
    .sum().reset_index())
```

```python
monthly_budget["Budget_Revenue"] = monthly_budget["Budget_Revenue"].astype(int)
```

```python
budget_vs_actual = monthly_budget.merge(
    actual_revenue,on=['Month', 'Product_ID'],how='left')
```

```python
budget_vs_actual['Actual_Revenue'] = (
    budget_vs_actual['Actual_Revenue'].fillna(0))
```

```python
budget_vs_actual['Variance'] = (
budget_vs_actual['Actual_Revenue']- budget_vs_actual['Budget_Revenue'])
```

```python
budget_vs_actual["Variance"] = budget_vs_actual["Variance"].astype(int)
```

```python
budget_vs_actual.head()
```

**(xx).Total Budget vs Total Actual**

```python
total_budget = budget_vs_actual['Budget_Revenue'].sum()
total_actual = budget_vs_actual['Actual_Revenue'].sum()

print("Total Budget:", total_budget)
print("Total Actual:", total_actual)
```

```python
total_variance = total_actual - total_budget

print("Total Variance:", total_variance)
```

**Conclusion**

- Actual revenue is lower than the budget.

**Insights**
- Total Budget: ₹3,975.35M
- Total Actual: ₹3,355.17M
- Variance: -₹620.17M

- Overall: The company achieved less revenue than planned, with an overall shortfall of about ₹620.17M.
